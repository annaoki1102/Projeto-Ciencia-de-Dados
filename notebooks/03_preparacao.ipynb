```python
import pandas as pd
```

```python
lista = []
for ano in range(2020, 2026):
    tabela = pd.read_csv(f"../data/raw/datatran{ano}.csv", sep=";", encoding="latin-1", decimal=",")
    lista.append(tabela)
df = pd.concat(lista, ignore_index=True)
df.shape
```

```python
dados = df[["data_inversa", "dia_semana", "horario", "uf", "causa_acidente", "classificacao_acidente", "mortos"]].copy()
dados.shape
```

```python
dados["data_inversa"] = pd.to_datetime(dados["data_inversa"])
dados["data_inversa"].dtype
```

```python
dados["hora"] = dados["horario"].str[:2].astype(int)
dados["hora"].sort_values().unique()
```

```python
dados["ano"] = dados["data_inversa"].dt.year
dados["ano"].value_counts().sort_index()
```

```python
dados["teve_morte"] = dados["mortos"] > 0
dados["teve_morte"].value_counts()
```

```python
dados["classificacao_acidente"] = dados["classificacao_acidente"].fillna("Não informado")
dados.isna().sum()
```

```python
dados.to_csv("../data/processed/acidentes_prf_2020_2025.csv", index=False)
```

## Conclusões da Fase 3: Preparação dos Dados

- Os 6 arquivos (2020 a 2025) foram juntados em uma tabela só.
- Foram mantidas 7 colunas: data_inversa, dia_semana, horario, uf, causa_acidente, classificacao_acidente e mortos.
- A coluna data_inversa foi convertida para data.
- Foi criada a coluna hora, com a hora cheia do acidente (de 0 a 23).
- Foi criada a coluna ano, tirada da data.
- Foi criada a coluna teve_morte, que é verdadeira quando mortos é maior que 0. Foram 29.147 acidentes com morte.
- Os 7 valores vazios de classificacao_acidente foram trocados por "Não informado".
- A tabela final tem 406.255 linhas e 10 colunas, sem valores vazios.
- O arquivo limpo foi salvo em data/processed/acidentes_prf_2020_2025.csv.
